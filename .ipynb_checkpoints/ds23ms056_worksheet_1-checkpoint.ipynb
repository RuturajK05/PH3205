```python
#Q1
import numpy as np
def is_prime(n):
    prime = True
    n_ = int(np.sqrt(n)) + 1
    for i in range(2,n_):
        if n%i == 0:
            prime = False
            break
    return prime
print(is_prime(9))
```

```python
#Q2
import numpy as np
def is_prime(n):
    prime = True
    n_ = int(np.sqrt(n)) + 1
    for i in range(2,n_):
        if n%i == 0:
            prime = False
            break
    return prime
a = 3
b = 27
for i in range(a,b+1):
    if is_prime(i) == True:
        print(i)
```

```python
#Q4
import numpy as np
def f(x,N):
    total = 0
    term = 1
    for i in range(1,N):
        total += term
        term = x*term/i
    return total 

y = [i for i in range(4,21,4)]
x = [i/10 for i in range(1,11)]

a = np.zeros((len(x), len(y)))
for i in range(len(x)):
    for j in range(len(y)):
        a[i,j] = f(x[i],y[j])
        
for i in range(len(x)):
    for j in range(len(y)):
        print(f"{a[i,j]:.6f}", end = " ")
    print('\n')
```

```python
#Q5
import matplotlib.pyplot as plt
#defining the functions

def leibnitz(N):
    total = 0
    for i in range(N):
        total += ((-1)**(i))/(2*i + 1)
    return total*4
def cont_fraction(n):
    total = 0
    step = 1
    while n > 0:
        if n != 1:
            step = 2 + ((2*n - 1)**2)/step
        else:
            step = 1/step
        n = n - 1
    total = 1 + step
    return 4/total

leibniz = []
cont_frac = []
#calculating for various n
x = [i for i in range(1,11)]
for i in x:
    print(f"With Leibniz --> {leibnitz(i):.8f}")
    leibniz.append(leibnitz(i))
    print(f"With Continued Fraction --> {cont_fraction(i):.8f}")
    cont_frac.append(cont_fraction(i))
#plotting
fig,ax = plt.subplots(figsize = (10,6))
ax.plot(x,leibniz, color = "blue", ls = ":", label = "Leibniz")
ax.plot(x, cont_frac, color = "red", ls = "dashdot", label = "Continued Fraction")
plt.legend()
plt.show()
```
