```python
#Q5
import matplotlib.pyplot as plt
#defining the functions

def leibnitz(N):
    total = 0
    for i in range(N):
        total += ((-1)**(i))/(2*i + 1)
    return total*4
def cont_fraction(n):
    total = 0
    step = 1
    while n > 0:
        if n != 1:
            step = 2 + ((2*n - 1)**2)/step
        else:
            step = 1/step
        n = n - 1
    total = 1 + step
    return 4/total

leibniz = []
cont_frac = []
#calculating for various n
x = [i for i in range(1,11)]
for i in x:
    print(f"With Leibniz --> {leibnitz(i):.8f}")
    leibniz.append(leibnitz(i))
    print(f"With Continued Fraction --> {cont_fraction(i):.8f}")
    cont_frac.append(cont_fraction(i))
#plotting
fig,ax = plt.subplots(figsize = (10,6))
ax.plot(x,leibniz, color = "blue", ls = ":", label = "Leibniz")
ax.plot(x, cont_frac, color = "red", ls = "dashdot", label = "Continued Fraction")
plt.legend()
plt.show()
```

```python
#Q1
#importing required modules
import numpy as np
import matplotlib.pyplot as plt

#defining the required function and its derivative
def f(x):
    return 3*(x) - np.tan(x)
    
def f_prime(x):
    return 3 - (np.cos(x)**-2)
    
#plotting the function on the graph
x = np.linspace(0,1.57,100)
plt.axhline(y = 0, color = "k", ls = "--")
plt.plot(x,f(x))
plt.show()

#defining our error margin and maximum iterations
eps = 1.0E-05
maxiter = 50

def bisection(f,L,R,eps,maxiter):
    #f: function whose roots we are to find
    #L: left interval
    #R: right interval
    c = (L+R)/2
    iteration = 1
    while abs(f(c)) >= eps and iteration < maxiter:
        if f(L)*f(R) > 0:
            print("Bad interval!")
            return None
        if f(L)*f(c) <= 0:
            R = c
        else:
            L = c
        c = (L+R)/2
        iteration += 1
        
    if iteration == maxiter and abs(f(c)) >= eps:
        print("Failed to converge!")
        return None
    else:
        return c, iteration

def secant(f,L,R,eps,maxiter):
    #f: function whose roots we are to find
    #L: left interval
    #R: right interval
    x2,x1 = R,L
    iteration = 1
    while abs(f(x2)) >= eps and iteration < maxiter:
        x3 = x2 - f(x2)*(x2 - x1)/(f(x2) - f(x1))
        x1 = x2
        x2 = x3
        iteration += 1
        
    if iteration == maxiter and abs(f(x2)) >= eps:
        print("Failed to converge!")
        return None
    else:
        return x2, iteration
    
def newton(f,x1,eps,maxiter):
    #f: function whose roots we are to find
    #x1: starting point
    x2 = x1
    iteration = 1
    while abs(f(x2)) >= eps and iteration < maxiter:
        x2 = x1 - (f(x1)/f_prime(x1))
        x1 = x2
        iteration += 1
        
    if iteration == maxiter and abs(f(x2)) >= eps:
        print("Failed to converge!")
        return None
    else:
        return x2, iteration
    
print(f"The root and number of iteration using Bisection method is {bisection(f,1.2,1.4, eps, maxiter)}") 
print(f"The root and number of iteration using Secant method is {secant(f,1.2,1.25, eps, maxiter)}") 
print(f"The root and number of iteration using Newton-Raphson method is {newton(f,1.2,eps,maxiter)}") 
```

```python
#Q2
#importing required modules
import numpy as np
import matplotlib.pyplot as plt

#defining the required functions
def V(x):
    return x**4 + (2*x/3)**3 - x**2
def kinetic(x):
    return -0.125 - V(x) #Kinetic energy is Total energy - Potential energy

#plotting the kinetic energy
x = np.linspace(-1.25,1.25,100)
plt.plot(x,kinetic(x))
plt.axhline(y = 0, color = "k", ls = "--")
plt.show()

#defining our error margin and maximum iterations
eps = 1.0E-05
maxiter = 100

def bisection(f,L,R,eps,maxiter):
    #f: function whose roots we are to find
    #L: left interval
    #R: right interval
    c = (L+R)/2
    iteration = 1
    while abs(f(c)) >= eps and iteration < maxiter:
        if f(L)*f(R) > 0:
            print("Bad interval!")
            return None
        if f(L)*f(c) <= 0:
            R = c
        else:
            L = c
        c = (L+R)/2
        iteration += 1
        
    if iteration == maxiter and abs(f(c)) >= eps:
        print("Failed to converge!")
        return None
    else:
        return c, iteration

def secant(f,L,R,eps,maxiter):
    #f: function whose roots we are to find
    #L: left interval
    #R: right interval
    x2,x1 = R,L
    iteration = 1
    while abs(f(x2)) >= eps and iteration < maxiter:
        x3 = x2 - f(x2)*(x2 - x1)/(f(x2) - f(x1))
        x1 = x2
        x2 = x3
        iteration += 1
        
    if iteration == maxiter and abs(f(x2)) >= eps:
        print("Failed to converge!")
        return None
    else:
        return x2, iteration
    
#printing the first root
print(f"The first root using Bisection method is {bisection(kinetic,-1.25,-0.5, eps, maxiter)}") 
print(f"The first root using Secant method is {secant(kinetic,-1.25,-1, eps, maxiter)}")

#printing the second root
print(f"The second root Bisection method is {bisection(kinetic,-0.5,0, eps, maxiter)}") 
print(f"The second root using Secant method is {secant(kinetic,-0.5,0, eps, maxiter)}")

#printing the third root
print(f"The third root Bisection method is {bisection(kinetic,0,0.7, eps, maxiter)}") 
print(f"The third root Secant method is {secant(kinetic,0,0.5, eps, maxiter)}")

#printing the fourth root
print(f"The fourth root Bisection method is {bisection(kinetic,0.7,1.25, eps, maxiter)}") 
print(f"The fourth root Secant method is {secant(kinetic,0.7,1.25, eps, maxiter)}")
```

```python
#Q1

#defining the functions
def f(x):
    return np.sin(x)

def forward_diff(f,x,h):
    return (f(x + h) - f(x))/h

def backward_diff(f,x,h):
    return (f(x) - f(x - h))/h

def central_diff(f,x,h):
    return (f(x + h) - f(x - h))/(2*h)

def fivepoint_diff(f,x,h):
    return (f(x - 2*h) - 8*f(x - h) + 8*f(x + h) - f(x + 2*h))/(12*h)

def line(x,m,c):
    return m*x + c

x = 2*np.pi/5
exact = np.cos(x)

methods = [forward_diff, backward_diff, central_diff, fivepoint_diff]
h_vals =  [0.5, 0.2, 0.1, 0.05, 0.02, 0.01, 0.005, 0.002, 0.001, 0.0005, 0.0002, 0.0001]
description = ["Forward","Backward","Central","Five Point"]
print("%10s %10s %10s %10s %10s %10s"%("h values","Forward","Backward","Central","Five Point","Exact"))
for h in h_vals:
    print("%10.6f"%(h), end = " ")
    for method in methods:
        print("%10.6f"%(method(f,x,h)), end = " ")
    print("%10.6f"%(exact))
i = 0
fig, ax = plt.subplots(figsize = (10,6))
for method in methods:
    if i == 3:
        y = [abs(method(f,x,h) - exact) for h in h_vals[:8]]
        ax.plot(np.log10(h_vals[:8]), np.log10(y[:8]),"o", label = description[i])
        popt,pcov = curve_fit(line, np.log10(h_vals[:8]), np.log10(y[:8]), p0=[1,1])
        ax.plot(np.log10(h_vals[:8]), line(np.log10(h_vals[:8]),*popt), color = "k", ls = "--")
        print(f"{description[i]} Slope ---> {round(popt[0],2)}")
        i+=1
    else:
        y = [abs(method(f,x,h) - exact) for h in h_vals]
        ax.plot(np.log10(h_vals), np.log10(y),"o", label = description[i])
        popt,pcov = curve_fit(line, np.log10(h_vals), np.log10(y), p0=[1,1])
        ax.plot(np.log10(h_vals), line(np.log10(h_vals),*popt), color = "k", ls = "--")
        print(f"{description[i]} Slope ---> {round(popt[0], 2)}")
        i+=1
plt.legend()
plt.title("Difference between Methods and Exact Value")
plt.xlabel("Log(h)")
plt.ylabel("Log(Difference)")
plt.show()
```

```python
#Q2
#defining the functions
def exp(x):
    return np.exp(x)

def trapezoid(f,x_o,x_n, m):
  N = 2*m + 1
  x = np.linspace(x_o,x_n,N)
  h = x[1] - x[0]
  return (h*(2*np.sum(f(x)) - f(x[0]) - f(x[-1]))/2)

def quadratic(f,x_o,x_n, m):
  N = 2*m + 1
  x = np.linspace(x_o,x_n,N)
  h = x[1] - x[0]
  return (h*(f(x[0]) + f(x[-1]) + 4*np.sum(f(x[1:-1:2])) + 2*np.sum(f(x[2:-1:2])))/3)

def cubic(f,x_o,x_n, m):
  N = 3*m + 1
  x = np.linspace(x_o,x_n,N)
  h = x[1] - x[0]
  return (3*h*(f(x[0]) + f(x[-1]) + 3*np.sum(f(x[1:-1])) - np.sum(f(x[3:-1:3])))/8)

def boole(f,x_o,x_n, m):
  N = 4*m + 1
  x = np.linspace(x_o,x_n,N)
  h = x[1] - x[0]
  return (2*h*(7*(f(x[0]) + f(x[-1])) + 32*np.sum(f(x[1:-1:2])) + 12*np.sum(f(x[2:-1:4])) + 14*np.sum(f(x[4:-1:4])))/45)

#printing the value of the integration of exponential function using different methods
print(f"Using Trapezoid Method --> {trapezoid(exp,0,1,5)}")
print(f"Using Simpson's 1/3 Method --> {quadratic(exp,0,1,5)}")
print(f"Using Simpson's 3/8 Method --> {cubic(exp,0,1,5)}")
print(f"Using Boole's Method --> {boole(exp,0,1,5)}")

exact = (np.exp(1) - 1)
x_0 = 0
x_N = 1
methods = [trapezoid, quadratic, cubic, boole]
m_vals =  [4, 8, 16, 32, 64]
description = ["Trapezoid", "Simpson's 1/3", "Simpson's 3/8", "Boole's"]
print("%15s %15s %15s %15s %15s %15s"%("m values","Trapezoid", "Simpson's 1/3", "Simpson's 3/8", "Boole's", "Exact"))
for m in m_vals:
    print("%15.f"%(m), end = " ")
    for method in methods:
        print("%15.8f"%(method(exp,x_0,x_N,m)), end = " ")
    print("%15.8f"%(exact))
i = 0
fig, ax = plt.subplots(figsize = (10,6))
for method in methods:
        y = [abs(method(exp,x_0,x_N,m) - exact) for m in m_vals]
        ax.plot(np.log10(m_vals), np.log10(y),"o", label = description[i])
        popt,pcov = curve_fit(line, np.log10(m_vals), np.log10(y), p0=[1,1])
        ax.plot(np.log10(m_vals), line(np.log10(m_vals),*popt), color = "k", ls = "--")
        print(f"{description[i]} Slope ---> {round(popt[0], 2)}")
        i+=1
plt.legend()
plt.title("Difference between Methods and Exact Value")
plt.xlabel("Log(m)")
plt.ylabel("Log(Difference)")
plt.show()
```

```python
#Q3

def f(x):
    return (x**(-2/3))*((1 - x)**(-1/3)) #has singularities at x = 0, and x = 1

def cubic(f,x_o,x_n, m):
    N = 3*m + 1
    x = np.linspace(x_o,x_n,N)
    h = x[1] - x[0]
    return (3*h*(f(x[0]) + f(x[-1]) + 3*np.sum(f(x[1:-1])) - np.sum(f(x[3:-1:3])))/8)

#We split the integral from 0 to 0.5 and 0.5 to 1
#For integration in the first half we make substitution p^3 = x, and for the second half we do p^2/3 = 1 - x, 
#and get respective functions f1 and f2, along with the proper limits of integration.

def f1(p):
    return 3*((1 - p**3)**(-1/3))

def f2(p):
    return 3*((1 - p**(3/2))**(-2/3))/2

x0 = 0
x1 = 0.5**(1/3)
x2 = 0.5**(2/3)

print(f"Vale after integrating using Simpson's 3/8 method --> {cubic(f1,x0,x1,50) + cubic(f2,x0,x2,50)}")
print(f"Exact value of integral --> {2*np.pi/np.sqrt(3)}")
```

```python
#Q1

#functions
def f(x,t):
    return -1*x*t

def euler(f, x, y, h):
    return y + h * f(x, y)

def midpoint(f, x, y, h):
    k1 = h * f(x, y)
    return y + (h * f(x + h / 2, y + k1 / 2))

def call(method, fn, y_ini, N, xs, h):
    y = y_ini
    ys = np.zeros((N, len(y_ini)))
    for i in range(N):
        x = xs[i]
        ys[i, :] = y
        y = method(fn, x, y, h)
    return ys

h = 1.0e-2
xf = 15

xs = np.arange(x0, xf, h)
N = len(xs) 

y_ini = np.asarray([1.0])  # Initial condition 

# Call the methods
y_euler = call(euler, f, y_ini, N, xs, h)
y_midpoint = call(midpoint, f, y_ini, N, xs, h)

# Generate exact values
y_exact = np.exp(-xs**2 / 2)

plt.figure(figsize=(6, 6))
plt.plot(xs, y_exact, color='blue',label='Exact', linewidth = 0.5)

plt.scatter(xs, y_euler, color='red', s=1, label='Euler')

plt.scatter(xs, y_midpoint, color='green', s=1, label='Midpoint')

plt.xlabel("x")
plt.ylabel("Solution")
plt.legend()
plt.title("Comparison of Numerical Methods")
plt.grid()
plt.show()

# Plotting errors
plt.figure(figsize=(6, 6))
plt.scatter(xs, [y[0] for y in y_euler] - y_exact, color='red', s=1, label='Euler')
plt.scatter(xs, [y[0] for y in y_midpoint] - y_exact, color='green', s=1, label='Midpoint')

plt.xlabel("x")
plt.ylabel("Errors")
plt.legend()
plt.title("Error Comparison of Numerical Methods")
plt.grid()
plt.show()


# Define the range for h
hs = 10.0 ** np.arange(-4.0, -2.0 + 0.2, 0.2)


# Calculate errors at the end
errors_euler = np.zeros(len(hs))
errors_midpoint = np.zeros(len(hs))
for i, h in enumerate(hs):
    xs = np.arange(x0, 10.0 + h, h)  # Generate xs up to 10.0 with step h
    N = len(xs)
    y_exact = np.exp(-xs**2 / 2)
    ys_euler = call(euler, f, y_ini, N, xs, h)
    ys_midpoint = call(midpoint, f, y_ini, N, xs, h)
    errors_euler[i] = abs(ys_euler[-1, 0] - y_exact[-1])
    errors_midpoint[i] = abs(ys_midpoint[-1, 0] - y_exact[-1])

# Error data in log-log format
xd = np.log10(hs)
yd_euler = np.log10(errors_euler)
yd_midpoint = np.log10(errors_midpoint)

# Define the fitting function
def line(x, m, c):
    return x*m + c

# Fit the data
popt_euler, _  = curve_fit(line, xd, yd_euler, p0=[1.0, 1.0])
popt_midpoint, _  = curve_fit(line, xd, yd_midpoint, p0=[1.0, 1.0])

# Generate smooth fit
xfine = np.linspace(xd[0], xd[-1], 1000)
yfine_euler = line(xfine, *popt_euler)
yfine_midpoint = line(xfine, *popt_midpoint)

# Plot the data and the fit
plt.figure(figsize=(8, 6))
plt.scatter(xd, yd_euler, color = "blue", label="Euler Data")
plt.scatter(xd, yd_midpoint, color = "red", label="Midpoint Data")
plt.plot(xfine, yfine_euler, label=f"Euler Fit. Slope = {popt_euler[0]:.2f}", color="k", ls = "--")
plt.plot(xfine, yfine_midpoint, label=f"Midpoint Fit. Slope = {popt_midpoint[0]:.2f}", color="k", ls = "--")
plt.xlabel(r"$\log_{10}$ of h")
plt.ylabel(r"$\log_{10}$ of Errors")
plt.grid()
plt.legend()
plt.title("Log-Log Error vs. Step Size")
plt.show()
```

```python
#Q2

#functions
def f(t, y):
    return np.asarray([y[1], -4*(np.pi**2)*y[0]])

def euler(f, x, y, h):
    return y + h * f(x, y)

def midpoint(f, x, y, h):
    k1 = h * f(x, y)
    return y + (h * f(x + h / 2, y + k1 / 2))

def call(method, fn, y_ini, N, xs, h):
    y = y_ini
    ys = np.zeros((N, len(y_ini)))
    for i in range(N):
        x = xs[i]
        ys[i, :] = y
        y = method(fn, x, y, h)
    return ys

h = 1.0e-2
x0 = 0.0
xf = 15

xs = np.arange(x0, xf, h)
N = len(xs)

y_ini = np.asarray([1.0, 0.0])  # Initial condition 

# Call the methods
y_euler = call(euler, f, y_ini, N, xs, h)
y_midpoint = call(midpoint, f, y_ini, N, xs, h)

# Plot for Euler method
plt.figure(figsize=(8, 10))
y_exact = np.cos(2*np.pi*xs)
plt.subplot(3, 1, 1)
plt.plot(xs,y_exact, label = "Exact Y", linewidth = 2)
plt.plot(xs, y_euler[:, 0], label="Y")
plt.plot(xs, y_euler[:, 1], label="P")
plt.xlabel("Time")
plt.title("Euler")
plt.legend()
plt.grid()

# Plot for Midpoint method
plt.subplot(3, 1, 2)
plt.plot(xs,y_exact, label = "Exact Y", linewidth = 2)
plt.plot(xs, y_midpoint[:, 0], label="Y")
plt.plot(xs, y_midpoint[:, 1], label="P")
plt.xlabel("Time")
plt.title("Midpoint")
plt.legend()
plt.grid()


# Define the range for h
hs = 10.0 ** np.arange(-4.0, -2.0 + 0.2, 0.2)


# Calculate errors of position
errors_euler = np.zeros(len(hs))
errors_midpoint = np.zeros(len(hs))
for i, h in enumerate(hs):
    xs = np.arange(x0, 10.0 + h, h)  
    N = len(xs)
    y_exact = np.cos(2*np.pi*xs) 
    ys_euler = call(euler, f, y_ini, N, xs, h)
    ys_midpoint = call(midpoint, f, y_ini, N, xs, h)
    errors_euler[i] = abs(ys_euler[-1, 0] - y_exact[-1])
    errors_midpoint[i] = abs(ys_midpoint[-1, 0] - y_exact[-1])

# Error data in log-log format
xd = np.log10(hs)
yd_euler = np.log10(errors_euler)
yd_midpoint = np.log10(errors_midpoint)

# Define the fitting function
def line(x, m, c):
    return x*m + c

# Fit the data
popt_euler, _  = curve_fit(line, xd, yd_euler, p0=[1.0, 1.0])
popt_midpoint, _  = curve_fit(line, xd, yd_midpoint, p0=[1.0, 1.0])

# Generate smooth fit
xfine = np.linspace(xd[0], xd[-1], 1000)
yfine_euler = line(xfine, *popt_euler)
yfine_midpoint = line(xfine, *popt_midpoint)

# Plot the data and the fit
plt.figure(figsize=(8, 6))
plt.scatter(xd, yd_euler, color = "blue", label="Euler Data")
plt.scatter(xd, yd_midpoint, color = "red", label="Midpoint Data")
plt.plot(xfine, yfine_euler, label=f"Euler Fit. Slope = {popt_euler[0]:.2f}", color="k", ls = "--")
plt.plot(xfine, yfine_midpoint, label=f"Midpoint Fit. Slope = {popt_midpoint[0]:.2f}", color="k", ls = "--")
plt.xlabel(r"$\log_{10}$ of h")
plt.ylabel(r"$\log_{10}$ of Errors")
plt.grid()
plt.legend()
plt.title("Log-Log Error vs. Step Size")
plt.show()
```

```python
#Q1

#functions 
def f(x, y): #acceleration function
    return -(4*np.pi**2) *y

def g(x,y):  #derivative function for euler and rk4
    return np.asarray([y[1], -4*(np.pi**2)*y[0]])

def euler(f, x, y, h):
    return y + h * f(x, y)

def verlet(f1, x, yn, ynm1, h):
    return 2*yn - ynm1 + f1(x, yn)*(h**2)

def velocity_verlet(f1, x, yn, vn, h):
    ynp1 = yn + vn*h + 0.5*f1(x, yn)*(h**2)
    vnp1 = vn + 0.5*(f1(x, yn) + f1(x + h, ynp1))*h
    return ynp1, vnp1

def rk4(f1, x, y, h):
    k1 = h * f1(x, y)
    k2 = h * f1(x + h/2, y + k1/2)
    k3 = h * f1(x + h/2, y + k2/2)
    k4 = h * f1(x + h, y + k3)
    return y + (k1 + 2*k2 + 2*k3 + k4)/6

def caller_y(method, fn, y0, v0, N, xs, h):
    ys = np.zeros(N)
    ys[0] = y0
    x0 = xs[0]
    ys[1] = y0 + v0*h + 0.5*fn(x0, y0)*(h**2) 
    for i in range(2, N):
        x = xs[i]
        ys[i] = method(fn, x, ys[i-1], ys[i-2], h)
    
    return ys


def caller_yv(method, fn, y0, v0, N, xs, h):
    
    ys = np.zeros(N)
    vs = np.zeros(N)
    ys[0] = y0
    vs[0] = v0
    for i in range(1, N):
        x = xs[i]
        ys[i], vs[i] = method(fn, x, ys[i-1], vs[i-1], h)
    
    return ys, vs

def caller_rk(method, fn, y_ini, N, xs, h):
    y = y_ini
    ys = np.zeros((N, len(y_ini)))
    for i in range(N):
        x = xs[i]
        ys[i, :] = y
        y = method(fn, x, y, h)
    return ys

h = 1.0e-2
x0 = 0.0
xf = 15

xs = np.arange(x0, xf, h)
N = len(xs)

y_ini = np.asarray([1.0, 0])  # Initial condition 
p_ini = np.asarray([0])

# Call the methods
y_v = caller_y(verlet, f, y_ini[0],p_ini[0], N, xs, h)
y_vv, p_vv = caller_yv(velocity_verlet, f, y_ini[0],p_ini[0], N, xs, h)
y_rk4 = caller_rk(rk4, g, y_ini, N, xs, h)

# Plot for Velocity method
plt.figure(figsize=(8, 10))
y_exact = np.cos(2*np.pi*xs)
plt.subplot(3, 1, 1)
plt.plot(xs, y_v, label="Y")
plt.plot(xs,y_exact, label = "Exact Y", ls = "--", color = "k")
plt.title("Verlet")
plt.legend()
plt.grid()

# Plot for Velocity Verlet method
plt.subplot(3, 1, 2)
plt.plot(xs, y_vv, label="Y")
plt.plot(xs, p_vv, label="P")
plt.plot(xs,y_exact, label = "Exact Y", ls = "--", color = "k")
plt.title("Velocity Verlet")
plt.legend()
plt.grid()

# Plot for RK4 method
plt.subplot(3, 1, 3)
plt.plot(xs, y_rk4, label="Y")
plt.plot(xs,y_exact, label = "Exact Y", ls = "--", color = "k")
plt.title("RK4")
plt.legend()
plt.grid()


# Define the range for h
hs = 10.0 ** np.arange(-4.0, -2.0 + 0.2, 0.2)


# Calculate errors of position
errors_v = np.zeros(len(hs))
errors_vv = np.zeros(len(hs))
errors_rk4 = np.zeros(len(hs))
for i, h in enumerate(hs):
    xs = np.arange(x0, 10.0 + h, h)  
    N = len(xs)
    y_exact = np.cos(2*np.pi*xs) 
    y_v = caller_y(verlet, f, y_ini[0],p_ini[0], N, xs, h)
    y_vv, p_vv = caller_yv(velocity_verlet, f, y_ini[0],p_ini[0], N, xs, h)
    y_rk4 = caller_rk(rk4, g, y_ini, N, xs, h)
    errors_v[i] = abs(y_v[-1] - y_exact[-1])
    errors_vv[i] = abs(y_vv[-1] - y_exact[-1])
    errors_rk4[i] = abs(y_rk4[-1, 0] - y_exact[-1])

# Error data in log-log format
xd = np.log10(hs)
yd_v = np.log10(errors_v)
yd_vv = np.log10(errors_vv)
yd_rk4 = np.log10(errors_rk4)

# Define the fitting function
def line(x, m, c):
    return x*m + c

# Fit the data
popt_v, _  = curve_fit(line, xd, yd_v, p0=[1.0, 1.0])
popt_vv, _  = curve_fit(line, xd, yd_vv, p0=[1.0, 1.0])
popt_rk4, _  = curve_fit(line, xd, yd_rk4, p0=[1.0, 1.0])

# Generate smooth fit
xfine = np.linspace(xd[0], xd[-1], 1000)
yfine_v = line(xfine, *popt_v)
yfine_vv = line(xfine, *popt_vv)
yfine_rk4 = line(xfine, *popt_rk4)

# Plot the data and the fit
plt.figure(figsize=(8, 6))
plt.scatter(xd, yd_v, color = "blue", label="Verlet Data")
plt.scatter(xd, yd_vv, color = "red", label="Verlet Velocity Data")
plt.scatter(xd, yd_rk4, color = "green", label="RK4 Data")
plt.plot(xfine, yfine_v, label=f"Verlet Fit. Slope = {popt_v[0]:.2f}", color="k", ls = "--")
plt.plot(xfine, yfine_vv, label=f"Verlet Velocity Fit. Slope = {popt_vv[0]:.2f}", color="k", ls = "--")
plt.plot(xfine, yfine_rk4, label=f"RK4 Fit. Slope = {popt_rk4[0]:.2f}", color="k", ls = "--")
plt.xlabel(r"$\log_{10}$ of h")
plt.ylabel(r"$\log_{10}$ of Errors")
plt.grid()
plt.legend()
plt.title("Log-Log Error vs. Step Size")
plt.show()
```

# Q2
The euqations of motions are as follows:
In the y-direction:
F = ma
ma_y = -mg -v_y*gamma
therfore a_y = dv_y/dt = -g -v_y*gamma/m

In the x-direction:
F = ma
ma_x = -v_x*gamma
therfore a_x = dv_x/dt = -v_x*gamma/m

```python
#Q2
grav = -9.8
gamma = 2
mass = 2

def f(t, Y): #derivative function
    x, y, vx, vy = Y
    return np.array([vx, vy, -vx*gamma/mass, grav - vy*gamma/mass])

x0 = 0
y0 = 0
vx0 = 5
vy0 = 5*np.sqrt(3)   # 10*sin(60°)

Y0 = np.array([x0, y0, vx0, vy0])

h = 0.01
tmax = 5.0   

def acc(vx, vy):
    ax = -vx*gamma/mass
    ay =  grav - vy*gamma/mass
    return ax,ay

def euler_step(f, t, y, h):
    return y + h*f(t,y)

def velocity_verlet_step(x, y, vx, vy, h):
    ax, ay = acc(vx, vy)

    x_new = x + vx*h + 0.5*ax*h*h
    y_new = y + vy*h + 0.5*ay*h*h

    vx_half = vx + 0.5*ax*h
    vy_half = vy + 0.5*ay*h

    ax_new, ay_new = acc(vx_half, vy_half)

    vx_new = vx_half + 0.5*ax_new*h
    vy_new = vy_half + 0.5*ay_new*h

    return x_new, y_new, vx_new, vy_new


def rk4_step(f, t, y, h):
    k1 = h*f(t,y)
    k2 = h*f(t+h/2, y+k1/2)
    k3 = h*f(t+h/2, y+k2/2)
    k4 = h*f(t+h, y+k3)
    return y + (k1 + 2*k2 + 2*k3 + k4)/6


def integrate(method):
    t = 0
    Y = Y0.copy()
    xs, ys = [], []

    while Y[1] >= 0:
        xs.append(Y[0])
        ys.append(Y[1])

        if method == "euler":
            Y = euler_step(f, t, Y, h)
        elif method == "rk4":
            Y = rk4_step(f, t, Y, h)
        elif method == "velocity_verlet":
            x,y,vx,vy = Y
            x,y,vx,vy = velocity_verlet_step(x,y,vx,vy,h)
            Y = np.array([x,y,vx,vy])
        t += h

    return np.array(xs), np.array(ys)

xe, ye = integrate("euler")
xv, yv = integrate("velocity_verlet")
xr, yr = integrate("rk4")
plt.figure(figsize=(8,6))
plt.plot(xe, ye, label="Euler")
plt.plot(xv, yv, label="Velocity Verlet")
plt.plot(xr, yr, label="RK4")
plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Projectile motion with air drag")
plt.legend()
plt.grid()
plt.show()

gamma_vals = np.arange(8,11)
fig,ax = plt.subplots(figsize = (10,6))
for gamma_val in gamma_vals:
    gamma = gamma_val
    xr, yr = integrate("rk4")
    ax.plot(xr, yr, label=f"Gamma = {gamma}")
    plt.legend()
plt.grid()
plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Projectile motion with air drag using RK4")
plt.show()
```

```python
#Q2

# Dormand-Prince
a2 = 1/5;  b21 = 1/5;
a3 = 3/10; b31 = 3/40;       b32 = 9/40;
a4 = 4/5;  b41 = 44/45;      b42 = -56/15;      b43 = 32/9;
a5 = 8/9;  b51 = 19372/6561; b52 = -25360/2187; b53 = 64448/6561; b54 =-212/729;
a6 = 1;    b61 = 9017/3168;  b62 = -355/33;     b63 = 46732/5247; b64 = 49/176;  b65 = -5103/18656;
a7 = 1;    b71 = 35/384;     b72 = 0;           b73 = 500/1113;   b74 = 125/192; b75 = -2187/6784;  b76 = 11/84;

c1 = 35/384; c2 = 0; c3 = 500/1113; c4 = 125/192; c5=-2187/6784; c6 = 11/84; c7 = 0;
c1s = 5179/57600; c2s = 0; c3s = 7571/16695; c4s = 393/640; c5s = -92097/339200; c6s = 187/2100; c7s = 1/40;

# Embedded Runge-Kutta formulas
def erk54h(f, x, y, h, k7bh):
    # Implement FSAL (first same as last)
    # The first
    # k1 = k7bh, unless k1 is zero (initially)
    if sum(k7bh)==0: 
        k1 = h*f(x, y)
    else:
        k1 = k7bh*h

    # Implement rest of the Butcher table
    k2 = h*f(x + a2*h, y + b21*k1)
    k3 = h*f(x + a3*h, y + b31*k1 + b32*k2)
    k4 = h*f(x + a4*h, y + b41*k1 + b42*k2 + b43*k3)
    k5 = h*f(x + a5*h, y + b51*k1 + b52*k2 + b53*k3 + b54*k4)
    k6 = h*f(x + a6*h, y + b61*k1 + b62*k2 + b63*k3 + b64*k4 + b65*k5)

    # get y fifth-order correct
    y5 = y + c1*k1 + c2*k2 + c3*k3 + c4*k4 + c5*k5 + c6*k6 # c7 = 0     
    
    # FSAL -> the last
    k7bh = f(x + a7*h, y5)

    # get y fourth-order correct
    y4 = y + c1s*k1 + c2s*k2 + c3s*k3 + c4s*k4 + c5s*k5 + c6s*k6 + c7s*k7bh*h

    # return y, error, k7bh
    err = abs(y5 - y4)
    return y5, err, k7bh

def caller54(fn, y_ini, x0, xT, h0, max_iter, abstol, reltol):
    y = np.asarray(y_ini); 
    h = h0;
    ys = np.zeros((max_iter, len(y_ini)))
    xs = np.zeros((max_iter, 1))
    xs[0] = x0 
    ys[0,:] = y
    x = x0
    i = 0 
    k = 1
    k7bh = np.zeros(len(y_ini))
    while x<=xT and i<max_iter:
        
        # calculate the tolerance
        tol = abstol + reltol*np.linalg.norm(ys[i,:])
        
        # get the next y and the error
        y, err, k7bh = erk54h(fn,xs[i],ys[i,:],h, k7bh)
        
        # get max error and the scale factor
        merr = np.max(err)
        if merr == 0.0: 
            merr = tol/100

        hnew = 0.9*h*(tol/merr)**0.2
        
        if merr<tol and h/5<hnew<5*h:   
            x += h
            i += 1
            xs[i] = x
            ys[i,:] = y
        
        # anyway we update h every time
        h = hnew
    
    # if the maximum number of steps exceeded
    if i>=max_iter:
        println(i," Increase max_iter.")
        return None
    print(f"The number of time steps required is {i}")
    return xs[1:i-1], ys[1:i-1,:]
```

```python
# Initialize parameters
max_iter = 20000
abstol = 1.0e-6
reltol = 1.0e-8
y_ini = [0.5,0]
t0 = 0.0
tf = 20.0
h0 = 1e-4

# Final call 
ts54, ys54 = caller54(g, y_ini, t0, tf, h0, max_iter, abstol, reltol)
```

```python
#functions
def rk4(f, x, y, k, h):
    k1 = h*f(x, y, k)
    k2 = h*f(x + h/2, y + k1/2, k)
    k3 = h*f(x + h/2, y + k2/2, k)
    k4 = h*f(x + h, y + k3, k)
    return y + (k1 + 2*k2 + 2*k3 + k4)/6

def caller_rk4(f, xlim, yini, k, N):
    x1, x2 = xlim
    xs = np.linspace(x1, x2, N)
    h = xs[1] - xs[0]
    y = yini
    ys = np.zeros((N, len(yini)))
    for i in range(N):
        ys[i] = np.array(y)
        y = rk4(f, xs[i], y, k, h)
    return xs, ys

def secant(ks, f, method, ybound, tol, maxiter):
    k1, k2 = ks
    iter = 0
    while abs(method(k2, f, ybound)) > tol and iter < maxiter:
        f1 = method(k1, f, ybound)
        f2 = method(k2, f, ybound)
        k1, k2 = k2, (f2*k1 - f1*k2)/(f2 - f1)
        iter += 1
    return iter, k2

def simp13(y, h):
    store = y[0]**2 + y[-1]**2
    for i in range(1, len(y)-1):
        if i % 2 == 0:
            store += 2*y[i]**2
        else:
            store += 4*y[i]**2
    return np.sqrt(store * h / 3)
```

```python
#Q1 a)

Vo = 40.0
Lo = 1.0

def pot_a(x):
    if abs(x) > Lo:
        return 0.0
    else:
        return -Vo*(1 - x**3)/2

def odefun_a(x, y, E):
    k2 = E - pot_a(x)
    return np.array([y[1], -k2 * y[0]])

def score_a(E, f, ybound):
    xlim = (-3.0, 3.0)
    N = 129
    yini1 = (ybound[0], 1e-2)
    xs1, ys1 = caller_rk4(odefun_a, xlim, yini1, E, N)
    a = abs(xs1 + Lo)
    n1 = np.where(a == np.min(a))[0][0]
    yini2 = (ybound[1], 1e-2)
    xs2, ys2 = caller_rk4(odefun_a, xlim[::-1], yini2, E, N)
    a = abs(xs2 + Lo)
    n2 = np.where(a == np.min(a))[0][0]
    ys2 = ys1[n1][0]*ys2/ys2[n2][0]
    return ys1[n1][1] - ys2[n2][1]
```

```python
xlim_a = (-3.0, 3.0)
ybound_a = (0.0, 0.0)

Es_test = np.linspace(-Vo, 0, 500)
scores_a = [score_a(E, odefun_a, ybound_a) for E in Es_test]

plt.figure(figsize=(10, 5))
plt.plot(Es_test, scores_a, 'b-')
plt.axhline(0, color='k', linestyle='--')
plt.xlabel('Energy')
plt.ylabel('Score')
plt.ylim(-10,10)
plt.grid()
plt.show()
```

```python
#function
def thomas_solve(a, b, c, d):
    """Thomas Algorithm to solve tridiagonal system A x = d."""
    n = len(b)
    # Forward elimination
    for i in range(1, n):
        w = a[i-1]/b[i-1]
        b[i]   = b[i]   - w*c[i-1]
        d[i]   = d[i]   - w*d[i-1]
    # Back substitution
    x = np.zeros(n)
    x[-1] = d[-1]/b[-1]
    for i in range(n-2, -1, -1):
        x[i] = (d[i] - c[i]*x[i+1])/b[i]
    return x
```

```python
#Q1 -> After some finite time t
L      = 1.0      # rod length
Nx     = 51
dx     = L/(Nx-1)
alpha  = 1e-4
dt     = 0.1
Nt     = 5000
r      = alpha*dt/dx**2

# Spatial grid
x = np.linspace(0, L, Nx)

# Dirichlet boundaries
T_left = 100.0
T_right= 100.0

# Uniform initial condition
T = np.full(Nx, 300.0)

# Construct A and B for interior nodes i=1..Nx-2
n_interior = Nx - 2
a_A = np.full(n_interior-1, -r/2)
b_A = np.full(n_interior,    1 + r)
c_A = np.full(n_interior-1, -r/2)

a_B = np.full(n_interior-1,  r/2)
b_B = np.full(n_interior,    1 - r)
c_B = np.full(n_interior-1,  r/2)

time_history = [T.copy()]

for n in range(Nt):
    T_old = T.copy()
    
    # 1) Form the RHS = B * T_old_interior + boundary terms
    T_in = T_old[1:-1]  # interior
    RHS = np.zeros(n_interior)

    # Multiply B*T_in
    for i in range(n_interior):
        RHS[i] = b_B[i]*T_in[i]
        if i > 0:
            RHS[i] += a_B[i-1]*T_in[i-1]
        if i < n_interior-1:
            RHS[i] += c_B[i]*T_in[i+1]

    # Boundary from time n => + (r/2)* T_left or T_right
    # Boundary from time n+1 => also + (r/2)* T_left, T_right on the RHS
    # Summed = r* T_left, r* T_right for the first/last interior row
    RHS[0]    += r * T_left     # i=1 sees boundary at x=0
    RHS[-1]   += r * T_right    # i=Nx-2 sees boundary at x=L

    # 2) Solve A * T_in^{new} = RHS
    sol = thomas_solve(a_A.copy(), b_A.copy(), c_A.copy(), RHS.copy())

    # 3) Put interior solution back + Dirichlet boundary
    T[1:-1] = sol
    T[0]    = T_left
    T[-1]   = T_right

    time_history.append(T.copy())

# Plot final vs initial
plt.plot(x, time_history[0], label='Initial')
plt.plot(x, T, label='Final')
plt.xlabel('x')
plt.ylabel('Temperature (K)')
plt.title('1D Heat Equation (Crank–Nicolson) with Gaussian IC')
plt.legend()
plt.show()
```

```python
#Q2 a)
###############################################################################
# 1. Physical / Numerical Parameters
###############################################################################
ħ       = 1.0      # Reduced Planck's constant (units)
m       = 1.0      # Particle mass (units)
Nx      = 2000     # Number of spatial grid points
x_min   = -20.0
x_max   =  20.0
L       = x_max - x_min
dx      = L / Nx

dt      = 0.025     # Time step
n_steps = 50     # Number of time steps
plot_every = 1   # Store/plot wavefunction every 'plot_every' steps

###############################################################################
# 2. Construct spatial and momentum grids
###############################################################################
x = np.linspace(x_min, x_max, Nx, endpoint=False)
dk = 2.0 * np.pi / L

# Create wave-number array:
#   indices 0..(Nx/2) increasing positive wavenumbers, 
#   indices (Nx/2+1).. Nx-1 negative wavenumbers, etc.
k_vals = np.fft.fftfreq(Nx, d=dx) * 2.0 * np.pi

###############################################################################
# 3. Define the Finite Square Well
###############################################################################
def finite_square_well(x, left_edge, right_edge, V0):
    """
    Returns a 1D potential array with a 'square well' from [left_edge, right_edge]
    of depth -V0 (i.e. negative potential).
    """
    V = np.zeros_like(x)
    mask = (x >= left_edge) & (x <= right_edge)
    V[mask] = V0
    return V

V0         = 40.0    # well depth
well_left  = 5.0
well_right =  7.0
V = finite_square_well(x, well_left, well_right, V0)

###############################################################################
# 4. Initial Gaussian Wave Packet
###############################################################################
x0    = -5.0   # initial center
sigma = 0.04
k0    = 3.0     # wave number -> momentum p = ħ*k0

def initial_gaussian_packet(x, x0, sigma, k0):
    """
    Returns a Gaussian wave packet centered at x0,
    with standard dev sigma, plane-wave factor e^{i k0 x}.
    """
    # Normalization factor is flexible; for large domains, we can skip it 
    # or just do approximate normalization.
    norm_factor = 1.0 / (np.sqrt(sigma * np.sqrt(np.pi)))
    return norm_factor * np.exp(-0.5 * ((x - x0)/sigma)**2) * np.exp(1j * k0 * x)

psi = initial_gaussian_packet(x, x0, sigma, k0)

###############################################################################
# 5. Split-Operator Step
###############################################################################
def split_operator_step(psi, V, dt, k_vals):
    """
    Performs one Strang-split time step:
      1) half-step kinetic
      2) full-step potential
      3) half-step kinetic
    psi     : current wavefunction in real space
    V       : potential array
    dt      : time step
    k_vals  : array of wave numbers
    """
    # 1) Half-step in kinetic (momentum) space
    psi_k = np.fft.fft(psi)
    # E_kin(k) = ħ^2 * k^2 / (2m)
    E_kin = (ħ**2 * k_vals**2) / (2.0 * m)
    # multiply by exp(-i E_kin dt/(2ħ))
    psi_k *= np.exp(-0.5j * E_kin * dt / ħ)
    psi = np.fft.ifft(psi_k)

    # 2) Full-step in potential (real) space
    # multiply by exp(-i V dt / ħ)
    psi *= np.exp(-1j * V * dt / ħ)

    # 3) Another half-step in kinetic
    psi_k = np.fft.fft(psi)
    psi_k *= np.exp(-0.5j * E_kin * dt / ħ)
    psi = np.fft.ifft(psi_k)

    return psi

###############################################################################
# 6. Main Time Loop
###############################################################################
snapshots = []
snapshots.append(psi.copy())  # store initial wavefunction

for step in range(1, n_steps + 1):
    psi = split_operator_step(psi, V, dt, k_vals)

    # Optionally check or normalize wavefunction:
    # norm_psi = np.sum(np.abs(psi)**2)*dx
    # psi /= np.sqrt(norm_psi)

    if step % plot_every == 0:
        snapshots.append(psi.copy())

###############################################################################
# 7. Animate / Save GIF
###############################################################################
# We'll use matplotlib.animation.FuncAnimation

fig, ax = plt.subplots()
line, = ax.plot(x, np.abs(snapshots[0])**2, lw=2)
ax.set_xlim(x_min, x_max)
ax.set_ylim(0, 1.1 * np.max(np.abs(snapshots[0])**2))
ax.set_xlabel("x")
ax.set_ylabel("|ψ|^2")

def update(frame):
    # frame is an index into snapshots
    psi_current = snapshots[frame]
    prob_density = np.abs(psi_current)**2
    line.set_ydata(prob_density)

    ax.set_ylim(0, 1.1 * np.max(prob_density))
    ax.set_title(f"Scattering Step: {frame*plot_every}")
    return line,

from matplotlib.animation import FuncAnimation

ani = FuncAnimation(fig, update, frames=len(snapshots), blit=False, interval=100);
HTML(ani.to_jshtml())
```

```python
#Q2 b)
###############################################################################
# 1. Physical / Numerical Parameters
###############################################################################
ħ       = 1.0      # Reduced Planck's constant (units)
m       = 1.0      # Particle mass (units)
Nx      = 2000      # Number of spatial grid points
x_min   = -20.0
x_max   =  20.0
L       = x_max - x_min
dx      = L / Nx

dt      = 0.025     # Time step
n_steps = 50     # Number of time steps
plot_every = 1   # Store/plot wavefunction every 'plot_every' steps

###############################################################################
# 2. Construct spatial and momentum grids
###############################################################################
x = np.linspace(x_min, x_max, Nx, endpoint=False)
dk = 2.0 * np.pi / L

# Create wave-number array:
#   indices 0..(Nx/2) increasing positive wavenumbers, 
#   indices (Nx/2+1).. Nx-1 negative wavenumbers, etc.
k_vals = np.fft.fftfreq(Nx, d=dx) * 2.0 * np.pi

###############################################################################
# 3. Define the Finite Square Well
###############################################################################
def squared_potential(x):
    """
    Returns a 1D potential array with a 'square well' from [left_edge, right_edge]
    of depth -V0 (i.e. negative potential).
    """
    V = np.zeros_like(x)
    
    return 0.1* (x**2)

V = squared_potential(x)

###############################################################################
# 4. Initial Gaussian Wave Packet
###############################################################################
x0    = -5.0   # initial center
sigma = 0.04
k0    = 3.0     # wave number -> momentum p = ħ*k0

def initial_gaussian_packet(x, x0, sigma, k0):
    """
    Returns a Gaussian wave packet centered at x0,
    with standard dev sigma, plane-wave factor e^{i k0 x}.
    """
    # Normalization factor is flexible; for large domains, we can skip it 
    # or just do approximate normalization.
    norm_factor = 1.0 / (np.sqrt(sigma * np.sqrt(np.pi)))
    return norm_factor * np.exp(-0.5 * ((x - x0)/sigma)**2) * np.exp(1j * k0 * x)

psi = initial_gaussian_packet(x, x0, sigma, k0)

###############################################################################
# 5. Split-Operator Step
###############################################################################
def split_operator_step(psi, V, dt, k_vals):
    """
    Performs one Strang-split time step:
      1) half-step kinetic
      2) full-step potential
      3) half-step kinetic
    psi     : current wavefunction in real space
    V       : potential array
    dt      : time step
    k_vals  : array of wave numbers
    """
    # 1) Half-step in kinetic (momentum) space
    psi_k = np.fft.fft(psi)
    # E_kin(k) = ħ^2 * k^2 / (2m)
    E_kin = (ħ**2 * k_vals**2) / (2.0 * m)
    # multiply by exp(-i E_kin dt/(2ħ))
    psi_k *= np.exp(-0.5j * E_kin * dt / ħ)
    psi = np.fft.ifft(psi_k)

    # 2) Full-step in potential (real) space
    # multiply by exp(-i V dt / ħ)
    psi *= np.exp(-1j * V * dt / ħ)

    # 3) Another half-step in kinetic
    psi_k = np.fft.fft(psi)
    psi_k *= np.exp(-0.5j * E_kin * dt / ħ)
    psi = np.fft.ifft(psi_k)

    return psi

###############################################################################
# 6. Main Time Loop
###############################################################################
snapshots = []
snapshots.append(psi.copy())  # store initial wavefunction

for step in range(1, n_steps + 1):
    psi = split_operator_step(psi, V, dt, k_vals)

    # Optionally check or normalize wavefunction:
    # norm_psi = np.sum(np.abs(psi)**2)*dx
    # psi /= np.sqrt(norm_psi)

    if step % plot_every == 0:
        snapshots.append(psi.copy())

###############################################################################
# 7. Animate / Save GIF
###############################################################################
# We'll use matplotlib.animation.FuncAnimation

fig, ax = plt.subplots()
line, = ax.plot(x, np.abs(snapshots[0])**2, lw=2)
ax.set_xlim(x_min, x_max)
ax.set_ylim(0, 1.1 * np.max(np.abs(snapshots[0])**2))
ax.set_xlabel("x")
ax.set_ylabel("|ψ|^2")

def update(frame):
    # frame is an index into snapshots
    psi_current = snapshots[frame]
    prob_density = np.abs(psi_current)**2
    line.set_ydata(prob_density)

    ax.set_ylim(0, 1.1 * np.max(prob_density))
    ax.set_title(f"Scattering Step: {frame*plot_every}")
    return line,

from matplotlib.animation import FuncAnimation

ani = FuncAnimation(fig, update, frames=len(snapshots), blit=False, interval=100);
HTML(ani.to_jshtml())
```

```python
#Q4
def take_a_step():
    s = np.random.rand()
    if s > 0.5:
        return 1
    else:
        return -1

def prob(n, N):
    P = 2 * np.exp(- n**2/(2*N))/np.sqrt(2 * np.pi * N)
    return P

num_walkers = 100000
num_step = 50

final_positions = np.zeros(num_step+1)
for walker in range(num_walkers):
    pos = 0
    for step in range(num_step):
        pos += take_a_step()
    pos_in_array = int((pos+num_step)/2.0)
    final_positions[pos_in_array] += 1

positions = np.linspace(-num_step, num_step, num_step+1)

plt.bar(positions, final_positions/num_walkers, label = "Data")
plt.plot(positions, prob(positions, num_step), color = "k", label = "Probability")
plt.legend()
plt.grid()
#we can see that there is perfect overlap between our theoretical probabilities and observational data.
```

```python
#Q5
def take_a_step():
    s = np.random.rand()
    if s > 0.5:
        return 1
    else:
        return -1

num_walkers = 100000
num_steps = 20

positions = np.zeros(num_walkers, dtype=int)
xaverage = np.zeros(num_steps)
for step in range(num_steps):
    xaverage[step] = sum([i**2 for i in positions])/num_walkers
    for walker in range(num_walkers):
        positions[walker] += take_a_step()
    
steps = [i for i in range(num_steps)]

plt.plot(steps,xaverage,'bo')
plt.xlabel("Steps")
plt.ylabel("<$x^2$>")
plt.grid()
```
